## Models Integrations With OpenAI , Google Gemini and Groq


```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GEMINIAI_API_KEY"] = os.getenv("GEMINIAI_API_KEY")
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python

from langchain.chat_models import init_chat_model

model = init_chat_model("gemini-3.5-flash",
                        model_provider="google_genai",
                        google_api_key=os.getenv("GEMINIAI_API_KEY")
                        )
                        
model
for chunk in model.stream("Explain about Langchain and Langgraph"):
    print(chunk.text)
```

```python
## Invoke the model
response = model.invoke("Hello how are you ")
```

## Streaming 

#### Most models can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves the user experience, perticularly for the longer responses. Calling stream() returns an iterator that yeilds output chunks as they are produced. You can use a loop to process each chunk in real-time:

```python
for chunk in model.stream("Explain about Langchain and Langgraph"):
    print(chunk.text , end="",flush=True)
```

## Batch


#### Batching is a collection of independent request to a model can significantly improve performance and reduce costs, as the processing can be done in parallel:



```python
responses = model.batch(
        [
        "What is AI?",
        "How do airplanes fly ?"
        ],
        config={
            'max_concurrency':5 #Limit to 5 parallel calls
        }   
    )
for response in responses:
    print(response.content)
```
