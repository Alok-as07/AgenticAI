```python
## Load the TXT file TextLoader

from langchain_community.document_loaders import TextLoader

loader = TextLoader("./files/textfiles/alok-about.txt",encoding="utf-8")
document = loader.load()
print(document)
```

```python
## Directory Loader for Txt files
from langchain_community.document_loaders import DirectoryLoader

dir_loader = DirectoryLoader(
    "./files/textfiles",
    glob="**/*.txt",
    loader_cls=TextLoader,
    show_progress=False
)
dir_loader.load()
```

```python
from langchain_community.document_loaders import DirectoryLoader,PyMuPDFLoader,PyPDFLoader

pdf_dir_loader = DirectoryLoader(
    "./files/pdffiles",
     glob="**/*.pdf",
     loader_cls=PyMuPDFLoader
)

pdf_dir_loader.load()
```
